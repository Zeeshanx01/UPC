# Recurssions:

````python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)


print(factorial(5))
````

# Project 1:

````python
import random

dict = {"r": 1, "p": -1, "s": 0}
reverseDict = {1: "Rock", -1: "Paper", 0: "Scissors"}
yourStr = input("Enter Your choice: \nr: Rock\np: Paper\ns: Scissors ")
if yourStr != "r" and yourStr != "p" and yourStr != "s":
    print("Invalid Input")
else:
    you = dict[yourStr]
    comp = random.choice([0, 1, -1])
    print(f"Computer chose: {reverseDict[comp]}\nYou chose: {reverseDict[you]}")
    if you == comp:
        print("it's a tie")
    else:
        if comp == -1 and you == 1:  # 0
            print("You loose!")
        elif comp == 1 and you == 0:  # 1
            print("You loose!")
        elif comp == 0 and you == -1:  # 1
            print("You loose!")
        elif comp == -1 and you == 0:  # -1
            print("You win!")
        elif comp == 1 and you == -1:  # 2
            print("You win!")
        elif comp == 0 and you == 1:  # -1
            print("You win!")
        else:
            print("Something went wrong")
````

# Project 1 (Short)

````python
import random

dict = {"r": 1, "p": -1, "s": 0}
reverseDict = {1: "Rock", -1: "Paper", 0: "Scissors"}
yourStr = input("Enter Your choice: \nr: Rock\np: Paper\ns: Scissors ")
if yourStr != "r" and yourStr != "p" and yourStr != "s":
    print("Invalid Input")
else:
    you = dict[yourStr]
    comp = random.choice([0, 1, -1])
    print(f"Computer chose: {reverseDict[comp]}\nYou chose: {reverseDict[you]}")
    if you == comp:
        print("it's a tie")
    else:
        if comp - you == -1 or comp - you == 2:
            print("You win!")
        else:
            print("You loose!")
````

````python
# Write file:
st = "Hello Zeeshan\nWelcome to Python Programming"
f = open("files/01_file.txt", "w")
f.write(st)
f.close()

# Read file:
f = open("files/01_file.txt")
data = f.read()
print(data)
f.close()

# Read lines:
f = open("files/01_file.txt")
lines = f.readlines()
print(lines, type(lines))
f.close()

# No need to close file using 'with' keyword:
with open("files/01_file.txt") as f:
    print(f.read())


# creating tables files
def generateTable(n):
    table = ""
    for i in range(1, 11):
        table += f"{n} x {i} = {n*i}\n"
    with open(f"tables/table_of_{n}.txt", "w") as f:
        f.write(table)


for i in range(2, 6):
    generateTable(i)
````
