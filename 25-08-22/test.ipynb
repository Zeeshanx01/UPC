
# Difference Between List and Tuple in Python
-----------------------------------------------------

1. Mutability:
   - List: Mutable (can be changed — add, remove, update elements).
   - Tuple: Immutable (cannot be changed after creation).

2. Syntax:
   - List: Defined with [ ] square brackets.
   - Tuple: Defined with ( ) round brackets.

3. Use Cases:
   - List: Best when data can change (dynamic collections).
   - Tuple: Best for fixed data (constants, keys in dictionaries, etc.).

# Python If-Elif-Else Example – Age Classification Program

```python
a = int(input("Enter your age: "))
if a >= 18:
    print("You are adult")
elif a >= 150:
    print("You are not human")
elif a < 18 & a > 12:
    print("You're a teenager")
else:
    print("You are a child")
```

# Age Checking Program Explanation
The program first takes the user's age as input and converts it into an integer. Then it checks different conditions: if the age is greater than or equal to 18, it prints "You are adult". If the age is 150 or more, it prints "You are not human". If the age is between 13 and 17, it prints "You're a teenager" (note: the code currently uses `&` which is a bitwise operator, it should be replaced with `and`). Finally, if none of the above conditions are true, the program prints "You are a child".

```python
# for loop:
for i in range(2):
    print(i)
print("--------------------")
for i in range(2, 5):
    print(i)
print("--------------------")

# while loop:
j = 0
while j < 3:
    print(j)
    j += 1
print("--------------------")

# another while loop:
l = [1, "Zeeshan", 3.2, False, "Hamza", 22, "Munir"]
k = 0
while k < len(l):
    print(l[k])
    k += 1
print("--------------------")

# another for loop:
for i in range(0, 40, 10):
    print(i)
print("--------------------")

# string
s = "Hello Zee!"
for i in s:
    print(i)
print("--------------------")

# for loop with else:
l = [1, 7, 11, 13]
for i in l:
    print(i)
else:
    print("done")

print("--------------------")
# break and continue
for i in range(5):
    if i == 3:
        print("Breaking the loop")
        break
    print(i)
print("--------------------")
for i in range(5):
    if i == 3:
        print("skipping the iteration")
        continue
    print(i)
print("--------------------")

# pass:
for i in range(20):
    pass
i = 0
while i < 4:
    print(i)
    i += 1
print("--------------------")
# star pattern:
n = 7
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    print("*" * (2 * i - 1), end="")
    print()
print("--------------------")

# box pattern:
for i in range(1, n + 1):
    if i == 1 or i == n:
        print("*   " * n, end="")
    else:
        print("*   ", end="")
        print("    " * (n - 2), end="")
        print("*   ", end="")
    print("")
```
