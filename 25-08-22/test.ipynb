-----------------------------------------------------
Difference Between List and Tuple in Python
-----------------------------------------------------

1. Mutability:
   - List: Mutable (can be changed — add, remove, update elements).
   - Tuple: Immutable (cannot be changed after creation).

2. Syntax:
   - List: Defined with [ ] square brackets.
   - Tuple: Defined with ( ) round brackets.

3. Use Cases:
   - List: Best when data can change (dynamic collections).
   - Tuple: Best for fixed data (constants, keys in dictionaries, etc.).

# Python Dictionary Example – Creating, Accessing, and Updating Data


```python
# Dictionary
user = {
    "name": "Zeeshan",
    "age": 22,
    "skills": ["Python", "Nextjs", "Tailwindcss"],
    "is_student": True,
}

print(f"{user}, \n{type(user)}")
print(user["name"])
#! print(user['name2'])
print(user.get("name2"))  # None
print(f"items: {user.items()}")
print(f"keys: {user.keys()}")
print(f"values: {user.values()}")
user.update(name="Zeeshan Munir", gmail="zeeshan1@gmail.com")
print(f"update: {user}")
```
