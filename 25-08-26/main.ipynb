# Inheritance:

```python
class Employee:
    company = "Google"
    name = "Zeeshan"
    age = 24

    def show(self):
        print(f"The name of the employee is {self.name} and the age is {self.age}.")


# * class Programmer:
# *     company = "Fiverr"

# *     def show(self):
# *         print(f"The name is {self.name} and the age is {self.age}.")

# *     def showLanguage(self):
# *         print(
# *             f"The name is {self.name} and he is fluent with this {self.language} language."
# *         )


class Programmer(Employee):
    company = "Fiverr"

    def showLanguage(self):
        print(
            f"The name is {self.name} and he is beginner in {self.language} language."
        )


a = Employee()
b = Programmer()
print(a.company, b.company)
```

# Multiple inheritance:

```python
class Employee:
    company = "Google"
    name = "Zeeshan"
    age = 24

    def show(self):
        print(f"The name of the employee is {self.name} and the age is {self.age}.")


class Coder:
    language = "Python"

    def printLanguages(self):
        print(f"The language is {self.language}")


class Programmer(Employee, Coder):
    company = "Fiverr"

    def showLanguage(self):
        print(
            f"The name is {self.name} and he is beginner in {self.language} language."
        )


a = Employee()
b = Programmer()
b.show()
b.printLanguages()
b.showLanguage()
```

# Multi-Level inheritance:

```python
class Employee:
    a = 1


class Manager(Employee):
    b = 2


class Programmer(Manager):
    c = 3


E = Employee()
M = Manager()
P = Programmer()
print(E.a)
#! print(E.b)
print(M.a, M.b)
print(P.a, P.b, P.c)
```

# More about inheritance:
- Super keyword
- Class methods
- Property decorators
- Operator overloading

```python
class Employee:
    def __init__(self):
        print("Constructor of Employee")

    a = 1

    @classmethod  # *
    # ? Always show directly class variable instead of instance variable
    def show(cls):
        print(f"Employee class variable a = {cls.a}")

    @property  # *
    def name(self):
        return f"{self.Fname} {self.Lname} {self.Lname}"

    @name.setter
    def name(self, value):
        self.Fname = value.split(" ")[0]
        self.Lname = value.split(" ")[1]


class Manager(Employee):
    def __init__(self):
        print("Constructor of Manager")

    b = 2


class Programmer(Manager):
    def __init__(self):
        super().__init__()  # *
        print("Constructor of Programmer")

    c = 3


# * Super
E = Employee()
print("-----")
M = Manager()
print("-----")
P = Programmer()
print("--------------------")

# * Class method
E.a = 45
E.show()
print(E.a)
print("--------------------")

# * Property decorators
E.name = "Zeeshan Munir"
print(E.name)
print(E.Fname, E.Lname)
print("--------------------")


class Num:
    def __init__(self, n):
        self.n = n


n = Num(1)
m = Num(2)

#! print(n + m)  # unsupported operand type(s) for +: 'Number' and 'Number'
print(n.n + m.n)


class Number:
    def __init__(self, n):
        self.n = n

    def __add__(self, num):
        return self.n + num.n


n = Number(1)
m = Number(2)

print(n + m)
```
